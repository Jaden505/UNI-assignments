### Evaluating truth tables

```python
def new_basic_table():
    table = {"p": ["T"] * 16 + ["F"] * 16,
             "q": (["T"] * 8 + ["F"] * 8) * 2,
             "r": (["T"] * 4 + ["F"] * 4) * 4,
             "s": (["T"] * 2 + ["F"] * 2) * 8,
             "t": ["T", "F"] * 16}

    return table
```

```python
def tte(inp):
    if inp == "T":
        return true
    elif inp == "F":
        return false
```

```python
def add_logical_statement(table, statement):
    row_entries = []
    symbols = list(statement.atoms())
    
    #remove these three lines to always show all five symbols
    irrelevant = [i for i in list(table.keys()) if i not in [str(i) for i in symbols]]
    for symbol in irrelevant:
        del table[symbol]    

    if len(symbols) == 2:
        sym_1, sym_2 = symbols[0], symbols[1]
        for i in range(32):
            row_entries.append(statement.subs({sym_1: tte(table[str(sym_1)][i]), sym_2: tte(table[str(sym_2)][i])}))

    elif len(symbols) == 3:
        sym_1, sym_2, sym_3 = symbols[0], symbols[1], symbols[2]
        for i in range(32):
            row_entries.append(statement.subs({sym_1: tte(table[str(sym_1)][i]),
                                               sym_2: tte(table[str(sym_2)][i]),
                                               sym_3: tte(table[str(sym_3)][i])}))

    elif len(symbols) == 4:
        sym_1, sym_2, sym_3, sym_4 = symbols[0], symbols[1], symbols[2], symbols[3]
        for i in range(32):
            row_entries.append(statement.subs({sym_1: tte(table[str(sym_1)][i]),
                                               sym_2: tte(table[str(sym_2)][i]),
                                               sym_3: tte(table[str(sym_3)][i]),
                                               sym_4: tte(table[str(sym_4)][i])}))

    elif (len(symbols)) == 5:
        sym_1, sym_2, sym_3, sym_4, sym_5 = symbols[0], symbols[1], symbols[2], symbols[3], symbols[4]
        for i in range(32):
            row_entries.append(statement.subs({sym_1: tte(table[str(sym_1)][i]),
                                               sym_2: tte(table[str(sym_2)][i]),
                                               sym_3: tte(table[str(sym_3)][i]),
                                               sym_4: tte(table[str(sym_4)][i]),
                                               sym_5: tte(table[str(sym_5)][i])}))

    table[statement] = row_entries

    return table
```

### Assignment step 1

Make truth tables for each formula. Count how often each formula is true and how often it is false (you can count by hand and pout answer in comment. Bonus point for counting using a python function that you define).


```python
def count_outcome_variation(statement):
    table = new_basic_table()
    table = add_logical_statement(table, statement)
    trues, falses = 0, 0
    
    for entry in table[statement]:
        if entry == true:
            trues += 1
        else:
            falses += 1
            
    print(f"Statement: {statement} has {trues} True outcomes and {falses} False outcomes.")
            
    return (trues, falses)

A = And( Implies(p, q), Implies(q, Not(r)), Or(p, r) )
B = Implies(Or(t, q), Or(Not(t), Not(s), r))
AndAB = And(A, B)
OrAB = Or(A, B)
ImpliesAB = Implies(A, B)
ImpliesBA = Implies(B, A)

count_outcome_variation(A)
count_outcome_variation(B)
count_outcome_variation(AndAB)
count_outcome_variation(OrAB)
count_outcome_variation(ImpliesAB)
count_outcome_variation(ImpliesBA)
```

### Assignment step 2

Can you comment on the result? Is it expected that some formules are more often true than others? 

Statement A has 8 True outcomes and 24 False outcomes, while Statement B has 28 True outcomes and 4 False outcomes. This indicates that Statement B is more likely to be true than Statement A. 

When combining both statements with an AND operation, the resulting statement has 7 True outcomes and 25 False outcomes, which is lower than either individual statement. This is expected because for an AND operation to be true, both statements must be true simultaneously, which is less likely.
Similarly, when combining both statements with an OR operation, the resulting statement has 29 True outcomes and 3 False outcomes, which is higher than either individual statement. This is also expected because for an OR operation to be true, at least one of the statements must be true, which is more likely.

Combining the statements with an implication operation (A implies B) results in 25 True outcomes and 7 False outcomes, while the reverse implication (B implies A) results in 11 True outcomes and 21 False outcomes. This asymmetry is expected because the truth of an implication depends on the relationship between the two statements, and one statement may be more likely to lead to the truth of the other.
